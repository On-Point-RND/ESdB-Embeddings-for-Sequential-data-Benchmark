```python
def get_run(number, specify="best", rewrite=False):
    path = Path(f"{log_path}/{dataset}/{method}/optuna/{number}")
    #print(pd.read_csv(path / "results.csv"))
    #print((path / "params.txt").read_text())
    save_path = Path(f"{configs_path}/specify/full/{dataset}/{method.lower()}")
    save_path.mkdir(parents=True, exist_ok=True)
    save_path = (save_path / f"{specify}.yaml")
    print(save_path)
    if not rewrite:
        assert not save_path.exists()
    save_path.write_text((path / "params.txt").read_text())

```

```python
def prepare_data(dataset, method):
    path = Path(f"{log_path}/{dataset}/{method}/optuna")
    df, study = optuna_df(path)
    col_to_drop = ["datetime_start", "datetime_complete", "system_attrs_fixed_params", "state", "value"]
    col_params = ["value", "duration"] + [col for col in df if "params_" in col]
    col_user = ["value", "duration"] + [col for col in df if "user" in col]
    df["duration"] = df["duration"].dt.total_seconds()
    return df, study, col_user, col_params


```

```python

methods = "NTP_GPT,NTP_GRU"

for dataset in all_datasets:
    for method in methods.split(','):
        df, study, col_user, col_params = prepare_data(dataset, method)
        if dataset == 'ett':
            df.rename(columns={'value': 'values_0'}, inplace=True)
        metric_id = 0
        for metric_id in range(4):
            try:
                value = [c for c in df.columns if 'value' in  c][metric_id]
                df_sorted = df[~df[value].isna()].sort_values(value, ascending=False).iloc[:, ]
                act_cols = [k for k in df_sorted.keys() if 'user_attrs_target' in k  and 'mean' in k]
                act_cols.append(value)
                param_cols = [k for k in df_sorted.keys() if 'params_model' in k]
                df_sorted_act_cols = df_sorted[act_cols].sort_values(value, ascending=False)
                numbers = pd.DataFrame(df_sorted_act_cols.iloc[0]).T
                metric_name = [c for c in numbers.columns if numbers[c].equals(numbers[f'values_{metric_id}'])][0]
                opt_metric = numbers[metric_name]
                opt_params = df[param_cols].iloc[df_sorted[act_cols].index[0]].rename(lambda x: x.replace('params_model.', '').replace('preprocess.', ''))
                task = ''
                #print(opt_metric.name)
                if 'accuracy' in opt_metric.name:
                    task = 'best_classification'
                else:
                    if 'anomaly' in opt_metric.name:
                        task = 'best_anomaly'
                    elif 'forecast' in opt_metric.name:
                        task = 'best_forecast'
                    elif 'reg' in opt_metric.name:
                        task = 'best_regression'
                #task = f'{dataset}_{method}_{task}'
                #dataset = f"{dataset}_full"
                print(f"{dataset}_{method}_{task}")
                get_run(int(opt_metric.index[0]), specify=task, rewrite=True)
            except:
                pass
                traceback.print_exc()
        print('\n')
```
